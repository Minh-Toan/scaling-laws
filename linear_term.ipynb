```python
import numpy as np
import matplotlib.pyplot as plt
import func
```

```python
def data_generate(d, k, n, Delta, sig, v):
    W0 = np.random.randn(k, d)/np.sqrt(d)
    X = np.random.randn(d, n)
    Z = np.random.randn(n)
    Y = sig(W0@X).T@v + np.sqrt(Delta)*Z

    mu0 = func.hermite_coeff(sig, 0)
    mu1 = func.hermite_coeff(sig, 1)
    Y_linear = mu1*(W0@X).T@v
    return W0, X, Y, Y_linear

def s_etimator(sig, Delta, v, Y, X):
    '''
    estimator for s = v.T W
    '''
    d = X.shape[0]
    mu0 = func.hermite_coeff(sig, 0)
    mu1 = func.hermite_coeff(sig, 1)
    Y0 = mu0*np.sum(v)
    s_hat = X@(Y-Y0)/mu1
    return s_hat


def align(u,v):
	return np.sum(u*v)/np.sqrt(np.sum(u**2)*np.sum(v**2))
```

```python
d = 200
gamma = 0.5
k = int(gamma*d)
alpha = 0.5
Delta = 0.01

n = int(alpha*d**2)

sig = lambda x: np.tanh(x)
# sig = lambda x: x + (x**2-1)/np.sqrt(2) + (x**3-3*x)/np.sqrt(6)
X = np.random.randn(d, n)

beta = 1
v = np.array([1/(p+1)**beta for p in range(k)])
v /= np.linalg.norm(v)

W0, X, Y, Y_linear = data_generate(d, k, n, Delta, sig, v)
s0 = W0.T@v

s_hat = s_etimator(sig, Delta, v, Y, X)
align(s_hat, s0)
```
